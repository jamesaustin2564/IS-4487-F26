## 2. Import data into Pandas

In this case we will load data from the statsmodels.org library

MTCARS is a traditional dataset used for learning.



```python
mtcars = sm.datasets.get_rdataset("mtcars", "datasets", cache=True).data
df = pd.DataFrame(mtcars)
df['carmodel'] = df.index
print(df)
```


## 4. Create basic charts

Pandas makes it easy to create data visualizations directly from a DataFrame using built-in charting functions. With simple commands like df.plot(), you can quickly generate line charts, bar graphs, histograms, and more without needing a lot of extra code. These visualizations help you see patterns, trends, and relationships in your data that might not be obvious from raw numbers alone. By learning how to use pandas’ plotting tools, you can turn your analysis into clear, visual insights that are easier to understand and communicate.

Below are examples of the following chart type:
- Line chart
- Bar chart
- Histogram
- Box plot
- Scatter plot




```python
#Line chart
df.plot(x='carmodel', y='mpg', kind='line', title="MPG by Car")
```

```python
#Bar chart
df.plot(x='carmodel', y='mpg', kind='bar', title="MPG by Car")
```

```python
#Histogram
df['mpg'].plot(kind='hist', bins=10, title="MPG Distribution")
```

```python
#Boxplot
df['mpg'].plot(kind='box', title = 'Miles Per Gallon')
```

```python
#Scatter plot
df.plot.scatter(x='mpg',y='hp',c='wt')
```

- 3 Create a scatter plot of wt (weight) vs. mpg.

```python
transmission_counts = df['am'].value_counts()
transmission_labels = {0: 'Automatic', 1: 'Manual'}
transmission_counts.index = transmission_counts.index.map(transmission_labels)

# Create the pie chart
transmission_counts.plot.pie(autopct='%1.1f%%', figsize=(6, 6), title='Distribution of Cars by Transmission Type')
```

```python
import matplotlib.pyplot as plt
import matplotlib.cm as cm

# Get the number of unique car models
num_models = len(df['carmodel'].unique())

# Generate a list of colors using a colormap (updated to address deprecation warning)
colors = plt.colormaps['viridis'](range(num_models)) # Using plt.colormaps for modern Matplotlib

# Create the bar chart with different colors for each car model
df.plot.bar(x='carmodel', y='hp', color=colors, figsize=(15, 7), title='Horsepower by Car Model')
plt.xlabel('Car Model')
plt.ylabel('Horsepower (hp)')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
df.plot.scatter(x='wt', y='mpg', title='Weight vs. Miles Per Gallon')
plt.xlabel('Weight (1000 lbs)')
plt.ylabel('Miles Per Gallon (mpg)')
plt.show()
```

```python
avg_mpg_by_cyl = df.groupby('cyl')['mpg'].mean()

avg_mpg_by_cyl.plot.bar(figsize=(8, 5), title='Average MPG by Number of Cylinders')
plt.xlabel('Number of Cylinders')
plt.ylabel('Average Miles Per Gallon (mpg)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
df['wt'].plot(kind='hist', bins=8, title='Distribution of Car Weight')
plt.xlabel('Weight (1000 lbs)')
plt.ylabel('Frequency')
plt.show()
```
